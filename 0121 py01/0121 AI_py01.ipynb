## Task 3 — Evaluate and plot variations of the Sine Function

Evaluate and plot  y=sin(x), y=sin(2x), and y=sin(x + \pi/2)  using **Python lists and NumPy arrays**.

Read the code carefully to review:

- Where loops are needed and where they are not
- How the same function can be implemented in different ways

```python
# Evaluating a function using a NumPy array and vectorized operations

import numpy as np
import math

# Create x values as a NumPy array over one period of the wave
x_array = np.linspace(0, 2 * math.pi, 50)

# Evaluate sin(x) using NumPy's vectorized function
y_sin_array = np.sin(x_array)

# Print a few values to verify the computation
print("First five x values:", x_array[:5])
print("First five sin(x) values:", y_sin_array[:5])
```

```python
# Evaluating a function using a Python list and a for loop

# Define a function that computes sin(2x) for a single value
def sin_2x(x):
    return math.sin(2 * x)

# Create x values as a Python list
x_list = list(np.linspace(0, 2 * math.pi, 50))

# Use a loop to evaluate the function for each x
y_sin_2x_list = []

for x in x_list:
    y_sin_2x_list.append(sin_2x(x))

# Print a few values to verify the computation
print("First five sin(2x) values:", y_sin_2x_list[:5])
```

```python
# Define a function that works on NumPy arrays
def sin_shifted(x):
    return np.sin(x + math.pi / 2)

# Reuse the NumPy x array
y_sin_shifted_array = sin_shifted(x_array)

# Print a few values to verify the computation
print("First five sin(x + pi/2) values:", y_sin_shifted_array[:5])
```

### Create simple plots to visualize your 3 functions

```python
import matplotlib.pyplot as plt

# Plot sin(x)
plt.figure()
plt.plot(x_array, y_sin_array)
plt.title("y = sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

# Plot sin(2x)
plt.figure()
plt.plot(x_list, y_sin_2x_list)
plt.title("y = sin(2x)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

# Plot sin(x + pi/2)
plt.figure()
plt.plot(x_array, y_sin_shifted_array)
plt.title("y = sin(x + π/2)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

### This is how you make a fancier plot with all your functions on it

```python
plt.figure(figsize=(9, 5))

plt.plot(x_array, y_sin_array,
         label="sin(x)",
         linewidth=2)

plt.plot(x_list, y_sin_2x_list,
         linestyle="--",
         label="sin(2x)",
         linewidth=2)

plt.plot(x_array, y_sin_shifted_array,
         linestyle=":",
         label=r"y=sin(2x)",
         linewidth=2)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Comparing Variations of sin(x)")
plt.legend()
plt.grid(True)

plt.savefig("sin_variations.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Task 4: Define and evaluate `g(x) = cos(x)`
1. Write a function `g(x)` that returns `cos(x)` using the `math` library.
2. Evaluate `g(x)` for every value in `x_list` using a loop.
3. Store results in `y_g_list`.

⭕️

```python
def g(x: float) -> float:
    """Returns cos(x) for a single number x (radians)."""
    return math.cos(x)

y_g_list = []

for x in x_list:
    y_g_list.append(g(x))

print("y_g_list:", y_g_list)
```

## Task 6: Linear vs Quadratic
Choose an `x` range and compute:
- A linear function (example form):  y=sin(2x)
- A quadratic function (example form): y=sin(x + \pi/2)

**Requirement**
- Use **lists + loops** for **one** of these.
- Use **NumPy arrays** for the **other**.

⭕️⭕️

Then **plot both** functions (your choice: same axes or separate).

⭕️⭕️

```python
# Choose your x-range
x_list2 = np.linspace(0, 100)

x_arr2 = np.array(x_list2)

# Pick coefficients (choose simple numbers)
a = 1.0
b = 2.0
c = -5000.0

def linear(x):
  return a * x + b

def quadratic(x):
  return a * x ** 2 + b * x + c

# Just so you know, this works too:
# y_linear_list = list(map(linear, x_list2))

y_linear_list = []
for x in x_list2:
  y_linear_list.append(linear(x))

y_quad_arr = quadratic(x_arr2)

# Plot linear function
plt.figure()
plt.plot(x_list2, y_linear_list)
plt.title("y=sin(2x)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

# Plot quadratic function
plt.figure()
plt.plot(x_list2, y_quad_arr)
plt.title("y=sin(x + \pi/2)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
